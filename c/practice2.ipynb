```python
chat_log = []
filtered = []
chat_log.append({
    "role": "user",
    "content": [{"text": "hello, how are you?"}]
})

chat_log.append({
    "role": "assistant",
    "content": [{"text": "I'm doing well, thank you!"}]
})

chat_log.append({
    "role": "user",
    "content": [{"text": "can you help me with Python?"}]
})

chat_log.append({
    "role": "assistant",
    "content": [{"text": "Of course! What do you need help with?"}]
})
count = 0
count1 =0
for message in chat_log:
    text = message["content"][0]["text"]
    r = message["role"]
    if "help" in text: 
         filtered.append(message)
         if "user" in r:
             count = count +1
             print(text + str(count))
         elif "assistant" in r:
              count1 = count1 +1
              print(text + str(count1))
        

print(filtered)
```
